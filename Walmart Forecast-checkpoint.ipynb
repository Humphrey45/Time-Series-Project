```python
waldata['Date'] = pd.to_datetime(waldata['Date'])
print(waldata.info())# to confirm the changes 
```

```python
# Date column converted. Now, i have to remove the Unnamed, month, year since i will be spliting Date column.
# I will also remove MarkDown columns because they are not needed in this analysis.
drop_columns = ['Unnamed: 0', 'Temperature', 'Fuel_Price','Unemployment', 'month', 'year', 'MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5', 'CPI']
waldata.drop(drop_columns, axis=1, inplace=True)
print(waldata.info())
```

```python
#Next step is to split the Date column in 'Day', 'Month', and 'year' and then drop the 'Date' column.
waldata['day'] = waldata['Date'].dt.day
waldata['month'] = waldata['Date'].dt.month
waldata['year'] = waldata['Date'].dt.year
waldata.drop('Date', axis=1, inplace=True)
print(waldata.info())
```

```python
# Change all column name to lowercase and remove whitespaces if any before performing Specfic Data Analysis (SDA)
waldata.columns = waldata.columns.str.lower() # For lowercase

# whitespaces removal
waldata.columns = waldata.columns.str.strip()
print(waldata.info())
```

```python
#Designed with the user in mind
# Viz. of wekkly sales based on holidays by the types of stores

plt.style.use('seaborn-poster')

wallabels = ['thanksgiving', 'super_bowl', 'labor_day', 'christmas'] # names for each bars

# This determines the length of each bar
mean_A = [27397.78, 20612.76, 20004.27, 18310.17]
mean_B = [18733.97, 12467.04, 12084.31, 11488.98]  
mean_C = [10179.27, 9893.46, 9696.56, 8031.52]

w = np.arange(len(wallabels)) #location of wallebels
width = 0.28 # this determines the size of the width

das, dx = plt.subplots(figsize=(18, 7)) # size of subplot considering user comfort

wal1 = dx.bar(w - width, mean_A, width, label='Type_A')
wal2 = dx.bar(w , mean_B, width, label='Type_B')
wal3 = dx.bar(w + width, mean_C, width, label='Type_C')

dx.set_ylabel('Weekly Average Sales')
dx.set_xticks(w)
dx.set_xticklabels(wallabels)
dx.legend()

# Function to attach and labels to the top of the bars to enable the user see and understand 
def barlabel(wal):
    for wals in wal:
        height = wals.get_height()
        dx.annotate('{}'.format(height),
                    xy=(wals.get_x() + wals.get_width() / 2, height),
                    xytext=(3, 6),
                    textcoords="offset points",
                    ha='center',va='bottom',
                    fontsize=12,
                   fontweight='bold',
                   color='black')

barlabel(wal1)
barlabel(wal2)
barlabel(wal3)

# Adding horizontal lines for holiday and non-holiday averages
plt.axhline(y=17094.30, color='red')  # holidays average
plt.axhline(y=15952.81, color='yellow') #no-holiday average

# Labeling the lines with their respective averages
plt.text(0.5, 17094.30, f'Holidays Avg: {17094.30}', color='black', fontsize=12, ha='right', va='bottom', fontweight='bold')
plt.text(0.5, 15952.82, f'No-Holiday Avg: {15952.82}', color='black', fontsize=12, ha='right', va='bottom', fontweight='bold')


das.tight_layout()

plt.legend()
plt.show()
```

Based on the data, the highest sales collectively comes from the super_bowel holidays while individually, type_A store had the highest sales across all holiday period.
